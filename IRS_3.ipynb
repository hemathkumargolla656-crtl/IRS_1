```python
import os
import string
from collections import defaultdict
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab') # Added to fix the LookupError

DOCS_FOLDER = "docs"
inverted_index = defaultdict(set)
stemmer = PorterStemmer() # Corrected typo from PorterStemter to PorterStemmer
stop_words = set(stopwords.words('english'))

# Create the DOCS_FOLDER if it doesn't exist, and create dummy files
if not os.path.exists(DOCS_FOLDER):
    os.makedirs(DOCS_FOLDER)
    # Create some dummy text files for demonstration
    with open(os.path.join(DOCS_FOLDER, 'doc1.txt'), 'w') as f:
        f.write('This is the first document. It talks about cats and dogs.')
    with open(os.path.join(DOCS_FOLDER, 'doc2.txt'), 'w') as f:
        f.write('The second document mentions dogs and birds.')
    with open(os.path.join(DOCS_FOLDER, 'doc3.txt'), 'w') as f:
        f.write('A third document about cats, birds, and fish.')

for filename in os.listdir(DOCS_FOLDER):
    if filename.endswith(".txt"):
        doc_id = filename
        with open(os.path.join(DOCS_FOLDER, filename), 'r', encoding='utf-8') as file:
            text = file.read().lower()
            text = text.translate(str.maketrans("", "", string.punctuation))
            tokens = word_tokenize(text)
            processed_tokens = [stemmer.stem(word) for word in tokens if word not in stop_words and word.isalpha()]

            for word in set(processed_tokens):
                inverted_index[word].add(doc_id)

print("Inverted Index (sample):")
for word in list(inverted_index.keys())[:10]:
    print(f"{word}: {sorted(list(inverted_index[word]))}")

print("\nTotal vocabulary size:", len(inverted_index))
```
